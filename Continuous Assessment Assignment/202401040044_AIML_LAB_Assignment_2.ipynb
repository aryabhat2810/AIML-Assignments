### **8. Breadth First Search (BFS)**

Breadth First Search (BFS) is an uninformed search algorithm that explores nodes level by level. It first visits all neighbouring nodes of the current node before moving to the next level.

BFS uses a **Queue (FIFO)** to store the nodes that need to be explored.

#### Example

Consider the disaster relief network:

**Warehouse → Camp A → Camp B → Camp C**

If the robot needs to reach Camp C, BFS explores the locations level by level:

**Warehouse → Camp A → Camp B → Camp C**

If multiple paths are available, BFS finds the path with the minimum number of edges in an unweighted graph.

#### Key Point

BFS is useful when we want to systematically explore all locations level by level and find a path with the minimum number of steps.

### **9. Depth First Search (DFS)**

Depth First Search (DFS) is an uninformed search algorithm that explores one path as deeply as possible before backtracking and trying another path.

DFS can be implemented using a **Stack (LIFO)** or recursion.

#### Example:

Suppose the relief network is:

**Warehouse → Camp A → Camp B → Camp C**

DFS may explore:

**Warehouse → Camp A → Camp B → Camp C**

If the path does not lead to the required destination, DFS backtracks to the previous location and explores another available path.

#### Key Point:

DFS is useful when we want to explore deeply into one possible route before checking other routes. However, the first route found may not be the shortest or lowest-cost route.

### **10. Depth Limited Search (DLS)**

Depth Limited Search (DLS) is a modified version of DFS in which a maximum depth is specified.

The algorithm explores nodes like DFS but stops when the predefined depth limit is reached.

#### Example:

Suppose the robot starts at:

**Warehouse → Camp A → Camp B → Camp C**

If the depth limit is **2**, the robot can explore:

**Warehouse → Camp A → Camp B**

but it will not explore Camp C because it is beyond the allowed depth.

If the depth limit is increased to **3**, Camp C can be reached.

####Key Point:

DLS is useful when we want to prevent the search from going too deep into the network. However, if the depth limit is too small, the required destination may not be found.

### **11. Iterative Deepening Search (IDS)**
Iterative Deepening Search (IDS) combines the ideas of DFS and DLS.

It repeatedly performs Depth Limited Search while increasing the depth limit step by step.

For example:

**Depth 0 → Depth 1 → Depth 2 → Depth 3 → ...**

The search continues until the destination is found.

#### Example:

Suppose the route is:

**Warehouse → Camp A → Camp B → Camp C**

IDS works as follows:

- Limit 0: Search only Warehouse.
- Limit 1: Search Warehouse → Camp A.
- Limit 2: Search Warehouse → Camp A → Camp B.
- Limit 3: Reach Camp C.

#### Key Point:

IDS is useful when the depth of the solution is not known in advance. It uses less memory than BFS while still systematically increasing the search depth.





### **12. Bidirectional Search**

Bidirectional Search performs two searches simultaneously:

- One search starts from the **source**.
- The other search starts from the **goal**.

The searches continue until they meet at a common node.

#### Example:

Suppose the relief route is:

**Warehouse → Camp A → Camp B → Camp C → Camp D**

Instead of searching only from Warehouse, Bidirectional Search can search:

**Forward:** Warehouse → Camp A → Camp B

and

**Backward:** Camp D → Camp C → Camp B

The searches meet at **Camp B**.

The complete route is then:

**Warehouse → Camp A → Camp B → Camp C → Camp D**

#### Key Point:

Bidirectional Search can reduce the amount of searching because the source and destination are explored simultaneously.

### **13. Greedy Best First Search**

Greedy Best First Search is an informed search algorithm that uses a **heuristic value** to decide which node appears closest to the destination.

It selects the node with the smallest estimated distance to the goal.

The evaluation function is:

**f(n) = h(n)**

where **h(n)** is the estimated cost from the current node to the goal.

#### Example:

Suppose the robot is at the Warehouse and has two possible camps:

- Camp A → estimated distance to goal = 5 km
- Camp B → estimated distance to goal = 3 km

Greedy Best First Search selects **Camp B** because it appears closer to the destination.

#### Key Point:

Greedy Best First Search can reach the goal quickly, but it does not consider the cost already travelled. Therefore, it may not produce the minimum-cost route.

### **14. A star Search**

A* Search is an informed search algorithm that considers both:

- The actual cost travelled so far.
- The estimated cost remaining to the destination.

It uses the evaluation function:

**f(n) = g(n) + h(n)**

where:

- **g(n)** = actual cost from the starting node to the current node.
- **h(n)** = estimated cost from the current node to the goal.
- **f(n)** = estimated total cost.

#### Example:

Suppose two possible routes are available:

**Route A:**  
Cost travelled = 6  
Estimated remaining cost = 2  
Total = 8

**Route B:**  
Cost travelled = 3  
Estimated remaining cost = 7  
Total = 10

A* compares the total values and selects **Route A** because:

**f(A) = 6 + 2 = 8**

**f(B) = 3 + 7 = 10**

#### Key Point:

A* is useful for disaster relief route planning because it considers both the distance already travelled and the estimated distance to the destination.


---



### **15. Hill Climbing**

Hill Climbing is a local search and optimization algorithm.

It starts with an initial solution and continuously moves to a neighbouring solution if that solution is better.

For the disaster relief problem, it can be used to optimize the order in which multiple camps are visited.

#### Example:

Suppose the initial delivery route is:

**Camp A → Camp B → Camp C → Camp D**

The algorithm changes the route slightly to create a neighbouring solution:

**Camp A → Camp C → Camp B → Camp D**

If the new route has a lower total travel cost, it becomes the new solution.

This process continues until no better neighbouring route is found.

#### Key Point:

Hill Climbing can quickly improve a solution, but it may get stuck at a **local optimum**, where no nearby solution is better even though a better solution may exist elsewhere.

###**16. Genetic Algorithm**

Genetic Algorithm (GA) is an optimization technique inspired by the process of natural selection.

It works with a **population of possible solutions** and improves them over multiple generations.

The main steps are:

1. **Initialize Population** – Generate several possible routes.
2. **Fitness Evaluation** – Calculate the quality or cost of each route.
3. **Selection** – Select better solutions as parents.
4. **Crossover** – Combine two parent solutions to create new solutions.
5. **Mutation** – Make small changes to introduce variation.
6. **Repeat** – Continue for several generations.
7. **Best Solution** – Select the best route obtained.

#### Example:

Suppose two possible delivery routes are:

**Parent 1:** Camp A → Camp B → Camp C → Camp D

**Parent 2:** Camp B → Camp D → Camp A → Camp C

The algorithm can combine parts of these routes through **crossover** to create new possible routes.

A mutation may then swap two camps to create another route.

After several generations, routes with lower travel costs are selected and improved.

#### Key Point:

Genetic Algorithm is useful when there are many possible delivery routes and finding the best route through simple exhaustive search would be difficult.

### **17. Constraint Satisfaction Problem (CSP) with Backtracking**

A Constraint Satisfaction Problem (CSP) is a problem in which a solution must satisfy a set of predefined **variables, domains, and constraints**.

In the disaster relief problem, CSP can be used to assign relief camps to delivery positions while satisfying conditions such as no repeated camp and a maximum travel cost.

#### Main Components

- **Variables:** Represent the decisions that need to be made.
- **Domain:** Contains the possible values that can be assigned to each variable.
- **Constraints:** Conditions that every valid solution must satisfy.

#### Example

Suppose three delivery positions need to be assigned:

**X1, X2, X3**

Possible camps:

**{Camp A, Camp B, Camp C}**

A constraint is that the same camp cannot be assigned more than once.

A possible valid assignment is:

**X1 = Camp A**  
**X2 = Camp B**  
**X3 = Camp C**

But:

**X1 = Camp A**  
**X2 = Camp A**

is invalid because Camp A is repeated.

#### Backtracking

If an assignment violates a constraint, the algorithm goes back to the previous decision and tries another value.

For example:

**X1 = Camp A → X2 = Camp A**

Since this violates the constraint, the algorithm backtracks and tries:

**X1 = Camp A → X2 = Camp B**

If this is valid, it continues to X3.

#### Key Point

CSP with Backtracking is useful when a disaster relief problem has multiple restrictions that must be satisfied while assigning camps, resources, schedules, or delivery positions.

### **18. Python Implementation**

```python
from collections import deque
import heapq
import math
import random
import time

START = "Warehouse"
GOAL = "Camp H"

GRAPH = {
    "Warehouse": {
        "Camp A": 2.5,
        "Camp D": 5.0
    },

    "Camp A": {
        "Warehouse": 2.5,
        "Camp B": 3.0,
        "Camp D": 3.0
    },

    "Camp B": {
        "Camp A": 3.0,
        "Camp C": 3.0,
        "Camp D": 4.0,
        "Camp E": 4.0,
        "Camp H": 8.0
    },

    "Camp C": {
        "Camp B": 3.0,
        "Camp F": 3.5
    },

    "Camp D": {
        "Warehouse": 5.0,
        "Camp A": 3.0,
        "Camp B": 4.0,
        "Camp E": 2.0,
        "Camp G": 2.5
    },

    "Camp E": {
        "Camp B": 4.0,
        "Camp D": 2.0,
        "Camp F": 3.0,
        "Camp G": 3.5,
        "Camp H": 4.0
    },

    "Camp F": {
        "Camp C": 3.5,
        "Camp E": 3.0,
        "Camp H": 3.2
    },

    "Camp G": {
        "Camp D": 2.5,
        "Camp E": 3.5,
        "Camp H": 3.5
    },

    "Camp H": {
        "Camp B": 8.0,
        "Camp E": 4.0,
        "Camp F": 3.2,
        "Camp G": 3.5
    }
}


COORDINATES = {
    "Warehouse": (0, 0),
    "Camp A": (2, 1),
    "Camp B": (5, 1),
    "Camp C": (8, 1),
    "Camp D": (3, 4),
    "Camp E": (5, 4),
    "Camp F": (8, 4),
    "Camp G": (3, 7),
    "Camp H": (7, 7)
}


def path_cost(path):
    total = 0

    for i in range(len(path) - 1):
        total += GRAPH[path[i]][path[i + 1]]

    return round(total, 2)


def reconstruct(parent, goal):
    path = []

    while goal is not None:
        path.append(goal)
        goal = parent[goal]

    return path[::-1]


def heuristic(node, goal=GOAL):

    x1, y1 = COORDINATES[node]
    x2, y2 = COORDINATES[goal]

    return math.sqrt(
        (x1 - x2) ** 2 +
        (y1 - y2) ** 2
    )


# =========================================================
# 1. BFS
# =========================================================

def bfs(start, goal):

    queue = deque([start])

    parent = {start: None}
    visited = {start}
    order = []

    while queue:

        current = queue.popleft()
        order.append(current)

        if current == goal:
            return reconstruct(parent, goal), order

        for neighbour in GRAPH[current]:

            if neighbour not in visited:

                visited.add(neighbour)
                parent[neighbour] = current
                queue.append(neighbour)

    return None, order


# =========================================================
# 2. DFS
# =========================================================

def dfs(start, goal):

    stack = [start]

    parent = {start: None}
    visited = {start}
    order = []

    while stack:

        current = stack.pop()
        order.append(current)

        if current == goal:
            return reconstruct(parent, goal), order

        for neighbour in reversed(
            list(GRAPH[current])
        ):

            if neighbour not in visited:

                visited.add(neighbour)
                parent[neighbour] = current
                stack.append(neighbour)

    return None, order


# =========================================================
# 3. DLS
# =========================================================

def dls(start, goal, limit):

    order = []

    def search(current, path):

        order.append(current)

        if current == goal:
            return path

        if len(path) - 1 >= limit:
            return None

        for neighbour in GRAPH[current]:

            if neighbour not in path:

                result = search(
                    neighbour,
                    path + [neighbour]
                )

                if result:
                    return result

        return None

    return search(start, [start]), order


# =========================================================
# 4. IDS
# =========================================================

def ids(start, goal, max_depth=10):

    total_order = []

    for depth in range(max_depth + 1):

        path, order = dls(
            start,
            goal,
            depth
        )

        total_order.extend(order)

        if path:
            return path, total_order, depth

    return None, total_order, None


# =========================================================
# 5. BIDIRECTIONAL SEARCH
# =========================================================

def bidirectional_search(start, goal):

    forward = deque([start])
    backward = deque([goal])

    forward_parent = {start: None}
    backward_parent = {goal: None}

    forward_visited = {start}
    backward_visited = {goal}

    order = []

    meeting = None

    while forward and backward:

        for _ in range(len(forward)):

            current = forward.popleft()
            order.append(current)

            if current in backward_visited:
                meeting = current
                break

            for neighbour in GRAPH[current]:

                if neighbour not in forward_visited:

                    forward_visited.add(neighbour)
                    forward_parent[neighbour] = current
                    forward.append(neighbour)

        if meeting:
            break

        for _ in range(len(backward)):

            current = backward.popleft()
            order.append(current)

            if current in forward_visited:
                meeting = current
                break

            for neighbour in GRAPH[current]:

                if neighbour not in backward_visited:

                    backward_visited.add(neighbour)
                    backward_parent[neighbour] = current
                    backward.append(neighbour)

        if meeting:
            break

    if meeting is None:
        return None, order

    left = reconstruct(
        forward_parent,
        meeting
    )

    right = []

    current = backward_parent[meeting]

    while current is not None:

        right.append(current)
        current = backward_parent[current]

    return left + right, order


# =========================================================
# 6. GREEDY BEST-FIRST SEARCH
# =========================================================

def greedy_best_first(start, goal):

    pq = []

    heapq.heappush(
        pq,
        (heuristic(start, goal), start)
    )

    parent = {start: None}
    visited = set()
    order = []

    while pq:

        _, current = heapq.heappop(pq)

        if current in visited:
            continue

        visited.add(current)
        order.append(current)

        if current == goal:
            return reconstruct(parent, goal), order

        for neighbour in GRAPH[current]:

            if neighbour not in visited:

                if neighbour not in parent:
                    parent[neighbour] = current

                heapq.heappush(
                    pq,
                    (
                        heuristic(
                            neighbour,
                            goal
                        ),
                        neighbour
                    )
                )

    return None, order


# =========================================================
# 7. A*
# =========================================================

def a_star(start, goal):

    pq = []

    heapq.heappush(
        pq,
        (heuristic(start, goal), start)
    )

    parent = {start: None}
    cost = {start: 0}

    visited = set()
    order = []

    while pq:

        _, current = heapq.heappop(pq)

        if current in visited:
            continue

        visited.add(current)
        order.append(current)

        if current == goal:
            return reconstruct(parent, goal), order

        for neighbour, distance in GRAPH[current].items():

            new_cost = (
                cost[current] + distance
            )

            if new_cost < cost.get(
                neighbour,
                float("inf")
            ):

                cost[neighbour] = new_cost
                parent[neighbour] = current

                f = (
                    new_cost +
                    heuristic(
                        neighbour,
                        goal
                    )
                )

                heapq.heappush(
                    pq,
                    (f, neighbour)
                )

    return None, order


# =========================================================
# MULTI-CAMP ROUTE COST
# =========================================================

CAMPS = [
    "Camp A",
    "Camp B",
    "Camp C",
    "Camp D",
    "Camp E",
    "Camp F",
    "Camp G"
]


def shortest_distance(start, goal):

    path, _ = a_star(start, goal)

    return path_cost(path)


def distribution_cost(route):

    complete_route = (
        [START] +
        route +
        [GOAL, START]
    )

    total = 0

    for i in range(len(complete_route) - 1):

        total += shortest_distance(
            complete_route[i],
            complete_route[i + 1]
        )

    return round(total, 2)


# =========================================================
# 8. HILL CLIMBING
# =========================================================

def hill_climbing(initial):

    current = initial[:]

    while True:

        current_cost = distribution_cost(
            current
        )

        best_route = current[:]
        best_cost = current_cost

        for i in range(len(current)):

            for j in range(i + 1, len(current)):

                neighbour = current[:]

                neighbour[i], neighbour[j] = (
                    neighbour[j],
                    neighbour[i]
                )

                cost = distribution_cost(
                    neighbour
                )

                if cost < best_cost:

                    best_route = neighbour
                    best_cost = cost

        if best_cost >= current_cost:
            return current

        current = best_route


# =========================================================
# 9. GENETIC ALGORITHM
# =========================================================

def fitness(route):

    return 1 / (
        distribution_cost(route) + 1
    )


def crossover(parent1, parent2):

    n = len(parent1)

    start, end = sorted(
        random.sample(range(n), 2)
    )

    child = [None] * n

    child[start:end + 1] = (
        parent1[start:end + 1]
    )

    remaining = [
        x for x in parent2
        if x not in child
    ]

    index = 0

    for i in range(n):

        if child[i] is None:

            child[i] = remaining[index]
            index += 1

    return child


def mutate(route):

    route = route[:]

    i, j = random.sample(
        range(len(route)),
        2
    )

    route[i], route[j] = (
        route[j],
        route[i]
    )

    return route


def genetic_algorithm(
    population_size=30,
    generations=50
):

    random.seed(42)

    population = []

    for _ in range(population_size):

        route = CAMPS[:]
        random.shuffle(route)

        population.append(route)

    best_route = None
    best_cost = float("inf")

    for _ in range(generations):

        population.sort(
            key=distribution_cost
        )

        if distribution_cost(
            population[0]
        ) < best_cost:

            best_route = population[0][:]
            best_cost = distribution_cost(
                best_route
            )

        new_population = population[:4]

        while len(new_population) < population_size:

            p1, p2 = random.sample(
                population[:15],
                2
            )

            child = crossover(p1, p2)

            if random.random() < 0.15:
                child = mutate(child)

            new_population.append(child)

        population = new_population

    return best_route, best_cost


# =========================================================
# 10. CSP - BACKTRACKING
# =========================================================

def csp_backtracking():

    variables = [
        "X1",
        "X2",
        "X3"
    ]

    assignment = {}

    def valid(variable, value):

        if value in assignment.values():
            return False

        if variable == "X1":

            return value in GRAPH[START]

        if variable == "X2":

            previous = assignment["X1"]

            return value in GRAPH[previous]

        if variable == "X3":

            previous = assignment["X2"]

            if value not in GRAPH[previous]:
                return False

            if GOAL not in GRAPH[value]:
                return False

            return True

        return False

    def backtrack(index):

        if index == len(variables):

            route = [
                START,
                assignment["X1"],
                assignment["X2"],
                assignment["X3"],
                GOAL
            ]

            if path_cost(route) <= 12:
                return route

            return None

        variable = variables[index]

        for value in GRAPH.keys():

            if valid(variable, value):

                assignment[variable] = value

                result = backtrack(
                    index + 1
                )

                if result:
                    return result

                del assignment[variable]

        return None

    return backtrack(0)


# =========================================================
# MAIN PROGRAM
# =========================================================

algorithms = [
    ("BFS", bfs),
    ("DFS", dfs)
]

print("=" * 60)
print("DISASTER RELIEF SUPPLY DISTRIBUTION ROBOT")
print("=" * 60)

for name, algorithm in algorithms:

    start_time = time.perf_counter()

    path, visited = algorithm(
        START,
        GOAL
    )

    end_time = time.perf_counter()

    print("\n", name)
    print("Path:", " -> ".join(path))
    print("Cost:", path_cost(path))
    print("Nodes:", len(visited))
    print(
        "Time:",
        round(
            (end_time - start_time) * 1000,
            4
        ),
        "ms"
    )


# DLS
path, visited = dls(
    START,
    GOAL,
    3
)

print("\nDLS")
print("Path:", " -> ".join(path))
print("Cost:", path_cost(path))
print("Nodes:", len(visited))


# IDS
path, visited, depth = ids(
    START,
    GOAL
)

print("\nIDS")
print("Path:", " -> ".join(path))
print("Cost:", path_cost(path))
print("Found at depth:", depth)


# Bidirectional
path, visited = bidirectional_search(
    START,
    GOAL
)

print("\nBidirectional Search")
print("Path:", " -> ".join(path))
print("Cost:", path_cost(path))


# Greedy
path, visited = greedy_best_first(
    START,
    GOAL
)

print("\nGreedy Best-First")
print("Path:", " -> ".join(path))
print("Cost:", path_cost(path))


# A*
path, visited = a_star(
    START,
    GOAL
)

print("\nA*")
print("Path:", " -> ".join(path))
print("Cost:", path_cost(path))


# Hill Climbing
initial = [
    "Camp A",
    "Camp B",
    "Camp C",
    "Camp D",
    "Camp E",
    "Camp F",
    "Camp G"
]

hill_route = hill_climbing(initial)

print("\nHill Climbing")
print("Route:", " -> ".join(hill_route))
print("Cost:", distribution_cost(hill_route))


# Genetic Algorithm
best_route, best_cost = genetic_algorithm()

print("\nGenetic Algorithm")
print("Best Route:", " -> ".join(best_route))
print("Best Cost:", best_cost)


# CSP
csp_route = csp_backtracking()

print("\nCSP - Backtracking")
print("Valid Route:", " -> ".join(csp_route))
print("Cost:", path_cost(csp_route))
```

###**19. Comparative Analysis**

**BFS vs DFS**

BFS systematically explores the search space level by level, while DFS follows a branch before backtracking. BFS can find a shallow solution, whereas DFS can use less memory but does not guarantee an optimal weighted route.

**DLS vs IDS**

DLS requires a predefined depth limit. IDS avoids choosing a single limit by repeatedly increasing the depth.

**Bidirectional Search**

Bidirectional Search reduces the search space by exploring from both the warehouse and the target camp.

Greedy Best-First vs A*

Greedy Best-First Search uses only:

$$ h(n) $$

while A* uses:

$$ g(n)+h(n) $$

Therefore, A* considers both the route already travelled and the estimated remaining route.

**Hill Climbing**

Hill Climbing improves a single candidate route by examining neighboring solutions. It is simple but may stop at a local optimum.

**Genetic Algorithm**

Genetic Algorithm evaluates multiple candidate routes simultaneously and uses selection, crossover and mutation to search for better distribution schedules.

**CSP**

CSP is appropriate when the robot must satisfy explicit requirements such as:

* carrying capacity,
* priority of emergency camps,
* route connectivity,
* maximum travel cost,
* delivery requirements.

###**20. Critical Analysis**
The experiment demonstrates that the disaster-relief distribution problem contains both route-finding and optimization components.

For a single relief camp, BFS, DFS, DLS, IDS and Bidirectional Search can be used to explore possible routes without heuristic information. Greedy Best-First and A* improve the search process by using estimated distance information.

For multiple relief camps, the problem becomes more complex because the robot must determine an efficient order of deliveries. Hill Climbing can improve an existing route but can become trapped at a local optimum. Genetic Algorithm provides multiple candidate solutions and searches for improved delivery sequences through successive generations.

Finally, CSP with Backtracking is useful when the route must satisfy explicit constraints. Thus, the experiment demonstrates that algorithm selection should depend on whether the requirement is route discovery, cost optimization or constraint satisfaction.
